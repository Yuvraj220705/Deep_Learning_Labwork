## Define Baseline CNN Model

### Subtask:
Define a Convolutional Neural Network (CNN) model without any regularization techniques (Dropout, L2, or Batch Normalization).


**Reasoning**:
I need to define a Convolutional Neural Network (CNN) model without any regularization techniques as per the subtask instructions. This involves sequentially adding layers to a Keras Sequential model.



```python
model_baseline = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_baseline.summary()
print("Baseline CNN model defined.")
```

**Reasoning**:
The previous execution failed with a `NameError` because `Sequential`, `Conv2D`, `MaxPooling2D`, `Flatten`, and `Dense` were not defined. This indicates that the necessary components from `tensorflow.keras.models` and `tensorflow.keras.layers` were not imported into the current scope. I will add the required import statements to the code block to resolve this issue.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model_baseline = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_baseline.summary()
print("Baseline CNN model defined.")
```

**Reasoning**:
Now that the baseline CNN model is defined, the next logical step according to typical machine learning workflow is to compile it. This involves specifying the optimizer, loss function, and metrics for training.



```python
model_baseline.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("Baseline CNN model compiled.")
```

## Train Baseline Model

```python
history_baseline = model_baseline.fit(x_train, y_train, epochs=10, validation_split=0.2)
print("Baseline CNN model training complete.")
```

## Define CNN Model with Dropout

### Subtask:
Define a Convolutional Neural Network (CNN) model by adding Dropout layers to the baseline architecture to mitigate overfitting. The Dropout layers will be added after the activation functions in the dense layers.


**Reasoning**:
I need to define a Convolutional Neural Network (CNN) model with Dropout layers as specified in the subtask. This involves creating a Sequential model and adding Dropout layers after the activation functions in the dense layers, and then printing its summary.



```python
from tensorflow.keras.layers import Dropout

model_dropout = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model_dropout.summary()
print("CNN model with Dropout defined.")
```

**Reasoning**:
The previous code block successfully defined the CNN model with Dropout. The next step is to compile this model, which involves setting the optimizer, loss function, and metrics before training, similar to how the baseline model was compiled.



```python
model_dropout.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("CNN model with Dropout compiled.")
```

**Reasoning**:
The CNN model with Dropout has been defined and compiled. The next step is to train it on the preprocessed MNIST training data, similar to the baseline model, to observe its performance.



```python
history_dropout = model_dropout.fit(x_train, y_train, epochs=10, validation_split=0.2)
print("CNN model with Dropout training complete.")
```
