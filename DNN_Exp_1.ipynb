```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X=np.random.randn(100,1)
Y=X*5
```

```python
def start_traning(weight):
  print(f"Creating A Network with this weight :{weight}")
  model=tf.keras.Sequential()
  model.add(tf.keras.Input(shape=(1,)))
  init=tf.keras.initializers.Constant(weight)

  for i in range(10):
    model.add(tf.keras.layers.Dense(1,activation='relu',kernel_initializer=init))

  with tf.GradientTape() as tape:
    prediction=model(X)
    # ✅ RIGHT: Build it (), then use it (Y, prediction)
    loss = tf.keras.losses.MeanSquaredError()(Y, prediction)

  grads=tape.gradient(loss,model.trainable_variables)
  for g in grads:
      print(g)
```

```python
start_traning(0.1)
```
